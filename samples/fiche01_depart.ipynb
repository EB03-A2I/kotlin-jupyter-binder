Executez d'abord la cellule ci-dessous : elle contient le code du cours (classes `Arme` et `Personnage`, fonction `attaquer`). Completez ensuite les cellules `TODO` avec `Shift+Entree`.

```kotlin
class Arme(val nom: String, val degats: Int)

class Personnage(
    val nom: String,
    var pv: Int,
    var arme: Arme? = null
)

fun Personnage.attaquer(cible: Personnage) {
    val degats = arme?.degats ?: 1
    cible.pv -= degats
    println("$nom attaque ${cible.nom} avec ${arme?.nom ?: "ses poings"} (-$degats PV)")
}

val heros = Personnage("Aldric", 20, Arme("Epee rouillee", 5))
val gobelin = Personnage("Gobelin", 8)
heros.attaquer(gobelin)
```
