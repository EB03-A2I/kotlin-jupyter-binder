Executez d'abord la cellule ci-dessous : elle contient le code du cours (`Personnage` avec bloc `init` et `companion object`, `Arme` en `data class`, singleton `ChroniqueDeCombat`). Completez ensuite les cellules `TODO` avec `Shift+Entree`.

```kotlin
data class Arme(val nom: String, val degats: Int)

class Personnage(
    val nom: String,
    var pv: Int,
    var arme: Arme? = null
) {
    init {
        require(pv >= 0) { "Les PV initiaux ne peuvent pas etre negatifs" }
    }
    companion object {
        const val PV_GOBELIN = 8
        fun creerGobelin(nom: String = "Gobelin", pv: Int = PV_GOBELIN, arme: Arme? = null) =
            Personnage(nom, pv, arme)
    }
}

fun Personnage.estVivant() = pv > 0

object ChroniqueDeCombat {
    var nombreAttaques = 0
        private set
    fun enregistrer(message: String) {
        nombreAttaques++
        println(message)
    }
}

fun Personnage.attaquer(cible: Personnage) {
    val degats = arme?.degats ?: 1
    cible.pv -= degats
    val message = "$nom attaque ${cible.nom} avec " +
        "${arme?.nom ?: "ses poings"} (-$degats PV)"
    ChroniqueDeCombat.enregistrer(message)
}

infix fun Personnage.soigne(montant: Int) {
    pv += montant
    println("$nom recupere $montant PV (${pv} PV)")
}

val heros = Personnage("Aldric", 20, Arme("Epee rouillee", 5))
val gobelin = Personnage.creerGobelin(pv = 6)
heros.attaquer(gobelin)
println("Attaques enregistrees : ${ChroniqueDeCombat.nombreAttaques}")
```
