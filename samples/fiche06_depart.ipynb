Executez d'abord la cellule ci-dessous : elle contient le code du cours (`map`, `filter`, `sumOf`, `maxByOrNull`). Completez ensuite les cellules `TODO` avec `Shift+Entree`.

```kotlin
interface Objet {
    val nom: String
    fun description(): String
}

data class Arme(override val nom: String, val degats: Int) : Objet {
    override fun description() = "$nom (+$degats degats)"
}

data class Bouclier(override val nom: String, val defense: Int) : Objet {
    override fun description() = "$nom (+$defense defense)"
}

class Personnage(
    val nom: String,
    var pv: Int,
    var arme: Arme? = null
) {
    val inventaire: MutableList<Objet> = mutableListOf()
    fun ramasser(objet: Objet) { inventaire.add(objet) }
}

val heros = Personnage("Aldric", 20)
heros.ramasser(Arme("Epee rouillee", 5))
heros.ramasser(Bouclier("Ecu de fer", 3))

val allie = Personnage("Elowen", 8, Arme("Dague", 3))
val groupe: List<Personnage> = listOf(heros, allie)

val noms: List<String> = groupe.map { it.nom }
println(noms)

val survivants: List<Personnage> = groupe.filter { it.pv > 10 }
println(survivants.map { it.nom })

val pvTotal: Int = groupe.sumOf { it.pv }
println("PV total du groupe : $pvTotal")

val plusRobuste: Personnage? = groupe.maxByOrNull { it.pv }
println("Le plus robuste : ${plusRobuste?.nom}")
```
