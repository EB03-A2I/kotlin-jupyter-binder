# Fiche 5 - Exercices : collections (List, MutableList) et lambdas

Executez d'abord la cellule ci-dessous : elle contient le code du cours (`Personnage.inventaire`, `ramasser`, `forEach` avec lambdas). Completez ensuite les cellules `TODO` avec `Shift+Entree`.

```kotlin
interface Objet {
    val nom: String
    fun description(): String
}

data class Arme(override val nom: String, val degats: Int) : Objet {
    override fun description() = "$nom (+$degats degats)"
}

data class Bouclier(override val nom: String, val defense: Int) : Objet {
    init { require(defense >= 0) { "La defense ne peut pas etre negative" } }
    override fun description() = "$nom (+$defense defense)"
}

open class Personnage(
    val nom: String,
    var pv: Int,
    var arme: Arme? = null
) {
    val inventaire: MutableList<Objet> = mutableListOf()
    fun ramasser(objet: Objet) {
        inventaire.add(objet)
    }
}

val heros = Personnage("Aldric", 20)
heros.ramasser(Arme("Epee rouillee", 5))
heros.ramasser(Bouclier("Ecu de fer", 3))
println("Objets transportes : ${heros.inventaire.size}")

heros.inventaire.forEach { objet -> println(objet.description()) }
heros.inventaire.forEach { println(it.description()) }   // parametre implicite

val allie = Personnage("Elowen", 15, Arme("Dague", 3))
val groupe: List<Personnage> = listOf(heros, allie)

var degatsPotentiels = 0
groupe.forEach { perso -> degatsPotentiels += perso.arme?.degats ?: 1 }
println("Degats potentiels du groupe : $degatsPotentiels")
```
