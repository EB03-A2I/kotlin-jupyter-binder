# Fiche 7 - Exercices : fonctions de portee (let, apply, also)

Executez d'abord la cellule ci-dessous : elle contient le code du cours (`apply`, `let`, `also`). Completez ensuite les cellules `TODO` avec `Shift+Entree`.

```kotlin
interface Objet {
    val nom: String
    fun description(): String
}
data class Arme(override val nom: String, val degats: Int) : Objet {
    override fun description() = "$nom (+$degats degats)"
}
data class Bouclier(override val nom: String, val defense: Int) : Objet {
    override fun description() = "$nom (+$defense defense)"
}
class Personnage(
    val nom: String,
    var pv: Int,
    var arme: Arme? = null
) {
    val inventaire: MutableList<Objet> = mutableListOf()
    fun ramasser(objet: Objet) { inventaire.add(objet) }
}

val heros = Personnage("Aldric", 20).apply {
    ramasser(Arme("Epee rouillee", 5))
    ramasser(Bouclier("Ecu de fer", 3))
}

heros.arme?.let { arme -> println("Arme equipee : ${arme.nom} (+${arme.degats})") }
    ?: println("Aucune arme equipee")

val allie = Personnage("Elowen", 8, Arme("Dague", 3)).also {
    println("Personnage cree : ${it.nom}")
}

println("Groupe : ${heros.nom}, ${allie.nom}")
```
