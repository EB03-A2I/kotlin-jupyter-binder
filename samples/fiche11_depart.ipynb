Executez d'abord la cellule ci-dessous : elle contient le code du cours (exceptions personnalisees, `try`/`catch`/`finally`, `require`). Completez ensuite les cellules `TODO` avec `Shift+Entree`.

```kotlin
interface Objet {
    val nom: String
    fun description(): String
}
data class Arme(override val nom: String, val degats: Int) : Objet {
    override fun description() = "$nom (+$degats degats)"
}
class Personnage(
    val nom: String,
    var pv: Int,
    var arme: Arme? = null
) {
    val inventaire: MutableList<Objet> = mutableListOf()
    fun ramasser(objet: Objet) { inventaire.add(objet) }
}

class ObjetIntrouvableException(message: String) : Exception(message)

fun Personnage.equiper(nomArme: String) {
    require(pv > 0) { "Un personnage vaincu ne peut pas s'equiper" }
    val objet = inventaire.find { it.nom == nomArme } as? Arme
        ?: throw ObjetIntrouvableException("Aucune arme nommee '$nomArme'")
    arme = objet
}

val heros = Personnage("Aldric", 20)
heros.ramasser(Arme("Epee rouillee", 5))

try {
    heros.equiper("Hache")
} catch (e: ObjetIntrouvableException) {
    println("Erreur : ${e.message}")
} finally {
    println("Tentative d'equipement terminee")
}

heros.equiper("Epee rouillee")
println("Arme equipee : ${heros.arme?.nom}")
```
