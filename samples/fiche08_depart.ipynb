Executez d'abord la cellule ci-dessous : elle contient le code du cours (`sealed class EtatPersonnage`, `when` exhaustif sans `else`). Completez ensuite les cellules `TODO` avec `Shift+Entree`.

```kotlin
interface Objet {
    val nom: String
    fun description(): String
}
data class Arme(override val nom: String, val degats: Int) : Objet {
    override fun description() = "$nom (+$degats degats)"
}
sealed class EtatPersonnage {
    object EnVie : EtatPersonnage()
    data class Etourdi(val toursRestants: Int) : EtatPersonnage()
    object Vaincu : EtatPersonnage()
}
class Personnage(
    val nom: String,
    var pv: Int,
    var arme: Arme? = null
) {
    var etat: EtatPersonnage = EtatPersonnage.EnVie
}
fun subirDegats(perso: Personnage, degats: Int) {
    perso.pv -= degats
    perso.etat = when {
        perso.pv <= 0 -> EtatPersonnage.Vaincu
        degats >= 10 -> EtatPersonnage.Etourdi(toursRestants = 2)
        else -> EtatPersonnage.EnVie
    }
}
fun decrireEtat(perso: Personnage): String = when (val etat = perso.etat) {
    is EtatPersonnage.EnVie -> "${perso.nom} est pret au combat"
    is EtatPersonnage.Etourdi -> "${perso.nom} est etourdi (encore ${etat.toursRestants} tour(s))"
    is EtatPersonnage.Vaincu -> "${perso.nom} est vaincu"
}

val heros = Personnage("Aldric", 20)
subirDegats(heros, 12)
println(decrireEtat(heros))

subirDegats(heros, 15)
println(decrireEtat(heros))
```
