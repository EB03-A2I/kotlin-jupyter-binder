# Fiche 10 - Exercices : delegation

Executez d'abord la cellule ci-dessous : elle contient le code du cours (delegation d'interface, `ObjetEnchante : Objet by base`). Completez ensuite les cellules `TODO` avec `Shift+Entree`.

```kotlin
interface Objet {
    val nom: String
    fun description(): String
}
data class Arme(override val nom: String, val degats: Int) : Objet {
    override fun description() = "$nom (+$degats degats)"
}
class Personnage(
    val nom: String,
    var pv: Int,
    var arme: Arme? = null
) {
    val inventaire: MutableList<Objet> = mutableListOf()
    fun ramasser(objet: Objet) { inventaire.add(objet) }
}

class ObjetEnchante(
    private val base: Objet,
    val bonus: String
) : Objet by base {
    override fun description() = "${base.description()} [$bonus]"
}

val heros = Personnage("Aldric", 20)
val epee = Arme("Epee rouillee", 5)
heros.ramasser(ObjetEnchante(epee, "flammes +3"))

val objet = heros.inventaire.first()
println(objet.nom)            // "Epee rouillee", venu de base via by
println(objet.description())  // redefinie, avec le bonus en plus
```
