```kotlin
interface Objet {
    val nom: String
    fun description(): String
}
data class Arme(override val nom: String, val degats: Int) : Objet {
    override fun description() = "$nom (+$degats degats)"
}
data class Bouclier(override val nom: String, val defense: Int) : Objet {
    override fun description() = "$nom (+$defense defense)"
}
class Personnage(
    val nom: String,
    var pv: Int,
    var arme: Arme? = null
) {
    val inventaire: MutableList<Objet> = mutableListOf()
    fun ramasser(objet: Objet) { inventaire.add(objet) }
}
fun <T : Objet> meilleurObjet(objets: List<T>, critere: (T) -> Int): T? =
    objets.maxByOrNull(critere)

inline fun <reified T : Objet> Personnage.compterType(): Int =
    inventaire.count { it is T }

val heros = Personnage("Aldric", 20)
heros.ramasser(Arme("Epee rouillee", 5))
heros.ramasser(Arme("Hache", 6))
heros.ramasser(Bouclier("Ecu de fer", 3))

val armes: List<Arme> = heros.inventaire.filterIsInstance<Arme>()
val meilleure = meilleurObjet(armes) { it.degats }
println("Meilleure arme : ${meilleure?.nom}")

println("Nombre d'armes : ${heros.compterType<Arme>()}")
println("Nombre de boucliers : ${heros.compterType<Bouclier>()}")
```
