# Fiche 4 - Exercices : interfaces, heritage et polymorphisme

Executez d'abord la cellule ci-dessous : elle contient le code du cours (interface `Objet` implementee par `Arme` et `Bouclier`, hierarchie `Personnage`/`Guerrier`). Completez ensuite les cellules `TODO` avec `Shift+Entree`.

```kotlin
interface Objet {
    val nom: String
    fun description(): String
}

data class Arme(override val nom: String, val degats: Int) : Objet {
    override fun description() = "$nom (+$degats degats)"
}

data class Bouclier(override val nom: String, val defense: Int) : Objet {
    init { require(defense >= 0) { "La defense ne peut pas etre negative" } }
    override fun description() = "$nom (+$defense defense)"
}

fun afficherObjet(o: Objet) = println(o.description())

open class Personnage(
    val nom: String,
    var pv: Int,
    var arme: Arme? = null
) {
    init { require(pv >= 0) { "Les PV initiaux ne peuvent pas etre negatifs" } }
    open fun crierDeGuerre() = "$nom se prepare au combat."
}

class Guerrier(nom: String, pv: Int, arme: Arme? = null) : Personnage(nom, pv, arme) {
    override fun crierDeGuerre() = "$nom hurle un cri de guerre assourdissant !"
}

afficherObjet(Arme("Epee rouillee", 5))
afficherObjet(Bouclier("Ecu de fer", 3))

val perso: Personnage = Guerrier("Aldric", 20, Arme("Epee rouillee", 5))
println(perso.crierDeGuerre())   // polymorphisme : la version de Guerrier est appelee
```
