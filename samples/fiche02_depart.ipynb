```kotlin
class Arme(val nom: String, val degats: Int)

class Personnage(
    val nom: String,
    var pv: Int,
    var arme: Arme? = null
)

fun creerGobelin(nom: String = "Gobelin", pv: Int = 8, arme: Arme? = null) = Personnage(nom, pv, arme)

fun Personnage.estVivant() = pv > 0

fun Personnage.attaquer(cible: Personnage) {
    val degats = arme?.degats ?: 1
    cible.pv -= degats
    println("$nom attaque ${cible.nom} avec ${arme?.nom ?: "ses poings"} (-$degats PV)")
}

infix fun Personnage.soigne(montant: Int) {
    pv += montant
    println("$nom recupere $montant PV (${pv} PV)")
}

val heros = Personnage("Aldric", 20, Arme("Epee rouillee", 5))
val gobelin = creerGobelin(pv = 4)
heros.attaquer(gobelin)
println("${gobelin.nom} vivant ? ${gobelin.estVivant()}")
```
